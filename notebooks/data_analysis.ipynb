```python
import pandas as pd

df= pd.read_csv('../data/raw/ssd.csv')
df.head()
```

```python
df= df.drop(['unnamed: 0'], inplace= True)
```

```python
num_cols= df.select_dtypes(include= ['int64', 'float64']).columns
num_cols
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

for cols in num_cols:
    for col2 in num_cols:
        if (cols != 'Unnamed: 0') and (col2 != 'Unnamed: 0') and (cols != col2):  
            plt.figure(figsize= (6,5))
            sns.scatterplot(data= df, x= cols, y= col2)
            plt.xlabel(cols)
            plt.ylabel(col2)
            plt.show()
```

```python
for cols in num_cols:
    if cols != 'Unnamed: 0':
        plt.figure(figsize= (5,6))
        sns.barplot(data= df, x= df['smoker'], y= cols)
        plt.xlabel('Smoker')
        plt.ylabel(cols)
        plt.show()
```
